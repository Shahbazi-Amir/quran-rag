```python
from pathlib import Path
import requests


# Project root
project_root = Path.cwd().parent if Path.cwd().name == "notebooks" else Path.cwd()


# Data directories
raw_dir = project_root / "data" / "raw"
processed_dir = project_root / "data" / "processed"

raw_dir.mkdir(parents=True, exist_ok=True)
processed_dir.mkdir(parents=True, exist_ok=True)


# Raw files paths
arabic_file = raw_dir / "quran_arabic_uthmani.txt"
fooladvand_file = raw_dir / "quran_fa_fooladvand.txt"
ansarian_file = raw_dir / "quran_fa_ansarian.txt"


print("Project root:", project_root)
print("Raw directory:", raw_dir)
print("Processed directory:", processed_dir)
```

### ساخت تابع دانلود

```python
def download_file(url, output_path):
    response = requests.get(url, timeout=60)
    response.raise_for_status()

    output_path.write_bytes(response.content)

    return len(response.content)


print("Download function is ready.")
```

```python
# Quran data sources

arabic_url = "https://tanzil.net/pub/download/download.php"
fooladvand_url = "https://tanzil.net/trans/fa.fooladvand"
ansarian_url = "https://tanzil.net/trans/fa.ansarian"


print("Sources are ready.")
print("Arabic:", arabic_url)
print("Fooladvand:", fooladvand_url)
print("Ansarian:", ansarian_url)
```

```python
# Download Quran translations

fooladvand_size = download_file(
    fooladvand_url,
    fooladvand_file
)

ansarian_size = download_file(
    ansarian_url,
    ansarian_file
)


print("Download completed")

print(f"Fooladvand: {fooladvand_size:,} bytes")
print(f"Ansarian: {ansarian_size:,} bytes")


print("\nFiles in raw folder:")
for file in raw_dir.iterdir():
    print(file.name)
```
